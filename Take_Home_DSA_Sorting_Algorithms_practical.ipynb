```python
# 1. Count the comparisons. Change bubble sort to return (sorted_list, comparisons).
#    Try it on an already sorted list of 1,000 numbers, then a reversed one. Explain the difference.


def bubble_count(items):
    items = items.copy()  # Keep the original list unchanged
    comparisons = 0

    n = len(items)

    for i in range(n - 1):
        swapped = False

        for j in range(n - 1 - i):
            comparisons += 1

            if items[j] > items[j + 1]:
                items[j], items[j + 1] = items[j + 1], items[j]
                swapped = True

        if not swapped:
            break

    return items, comparisons


_, c1 = bubble_count(list(range(1000)))
_, c2 = bubble_count(list(range(1000, 0, -1)))

assert c1 == 999
assert c2 == 499500

print("✅ bubble_count passes:", c1, c2)
```

The already sorted list requires only 999 comparisons because Bubble Sort makes one pass, finds no swaps and stops early.

The reversed list requires 499,500 comparisons because the algorithm must perform all its passes to sort the numbers.

 This demonstrates the difference between the best-case time complexity, O(n), and worst-case time complexity, O(n²), of optimized Bubble Sort

```python
# 2. Sort these M-Pesa style transactions by amount, largest first,
#    and keep transactions with the same amount in their original order.
transactions = [("TX1", 5000), ("TX2", 20000), ("TX3", 5000), ("TX4", 750)]

def sort_transactions(tx):
     return sorted(tx, key=lambda t: t[1], reverse=True)

assert sort_transactions(transactions) == [("TX2", 20000), ("TX1", 5000), ("TX3", 5000), ("TX4", 750)]
print("✅ sort_transactions passes")
```

### Explanation

The function sorts M-Pesa transactions according to their amounts, starting with the largest and ending with the smallest. It uses Python's `sorted()` function and `key=lambda t: t[1]` to select the transaction amount for sorting. `reverse=True` arranges the transactions in descending order.

The sorting is stable, meaning transactions with equal amounts maintain their original order. TX1 therefore remains before TX3 because both have an amount of 5,000.

The assertion checks whether the function returns the expected result. If the test passes, the program prints a success message.
