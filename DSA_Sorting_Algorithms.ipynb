## Section 1 · Bubble Sort
Compare neighbours and swap them if they are in the wrong order. Each pass pushes the largest remaining value to the end.
Complete the swap, then run. The `print` shows the list after every pass: does it match your paper?

```python

def bubble_sort(items):
    a = items[:]  # Work on a copy
    n = len(a)

    for i in range(n - 1):
        swapped = False

        for j in range(n - 1 - i):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True

        print("after pass", i + 1, a)

        if not swapped:
            break

    return a

assert bubble_sort([5, 1, 4, 2, 8]) == [1, 2, 4, 5, 8]
print("✅ bubble_sort passes")
```

**Remember**

*The key idea of Bubble Sort is that larger elements gradually move towards the end of the list through repeated comparisons and swaps.*


Bubble Sort is a sorting algorithm that arranges elements in ascending order by repeatedly comparing neighbouring elements and swapping them when they are in the wrong order.


In this implementation, the function first creates a copy of the original list. It then compares adjacent elements. If the first element is greater than the second, they are swapped. This process continues until the largest elements gradually move to the end of the list.


The `swapped` variable helps improve the algorithm's performance. If no swaps occur during a pass, it means the list is already sorted, so the algorithm stops early.

The `assert` statement checks whether the function produces the expected sorted list.


**Conclusion:** *Bubble Sort successfully arranges the list in ascending order while using an early-stopping technique to avoid unnecessary comparisons.*

## Section 2 · Selection Sort
Find the smallest value in the unsorted part and swap it to the front of that part.

```python
def selection_sort(items):
    a = items[:]
    n = len(a)
    for i in range(n - 1):
        smallest = i
        for j in range(i + 1, n):
            # if a[j] is smaller than a[smallest], remember j
          if a[j] < a [smallest]:
            smallest = j
        a[i], a[smallest] = a[smallest], a[i]
        print("after step", i + 1, a)
    return a

assert selection_sort([29, 10, 14, 37, 13]) == [10, 13, 14, 29, 37]
print("✅ selection_sort passes")
```

Selection Sort is a sorting algorithm that arranges elements in ascending order by repeatedly finding the smallest element in the unsorted part of the list.

In this implementation, the `smallest` variable stores the position of the smallest element. The inner loop compares the remaining elements to find a smaller value. If one is found, its position is updated.

After finding the smallest element, the algorithm swaps it with the element at the current position. This process continues until the entire list is sorted.

The `assert` statement checks whether the function produces the expected sorted list.

## Section 3 · Insertion Sort
Like sorting cards in your hand: take the next card and slide it left until it is in place.

```python
def insertion_sort(items):
    a = items[:]
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        # while j >= 0 and a[j] > key: shift a[j] one place right, move j left
        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = key
        print("after inserting", key, a)
    return a

assert insertion_sort([5, 2, 4, 6, 1, 3]) == [1, 2, 3, 4, 5, 6]
print("✅ insertion_sort passes")
```


Insertion Sort is a sorting algorithm that arranges elements in ascending order by taking one element at a time and inserting it into its correct position.


In this implementation, the algorithm starts with the second element and compares it with the elements before it. If the previous elements are greater, they are shifted one position to the right to create space for the current element.


The `key` variable stores the element being inserted, while the `j` variable helps move backwards through the list to find the correct position.

This process continues until all elements are arranged in the correct order.

The `assert` statement checks whether the function produces the expected sorted list.

## Section 4 · Merge Sort
1. Write `merge`: combine two **already sorted** lists into one sorted list.
2. Then `merge_sort` uses recursion: split in half, sort each half, merge.

```python
def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        # take the smaller front item from left or right
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])      # whatever is left over
    result.extend(right[j:])
    return result

assert merge([3, 27, 38, 43], [9, 10, 82]) == [3, 9, 10, 27, 38, 43, 82]
print("✅ merge passes")
```



The `merge()` function combines two already sorted lists into one sorted list.

It starts by creating an empty list called `result` and uses two variables, `i` and `j`, to track the positions of elements in the left and right lists.

The function compares the first available elements from both lists and adds the smaller element to the result list. It continues comparing and adding elements until one of the lists is empty.

The `extend()` method then adds any remaining elements from either list to the result.

Finally, the function returns one combined sorted list. The `assert` statement checks whether the function produces the expected output.

```python
def merge_sort(items):
    if len(items) <= 1:                 # base case
        return items
    mid = len(items) // 2
    left = merge_sort(items[:mid])      # recursive case
    right = merge_sort(items[mid:])
    return merge(left, right)

assert merge_sort([38, 27, 43, 3, 9, 82, 10]) == [3, 9, 10, 27, 38, 43, 82]
print("✅ merge_sort passes")
```



Merge Sort is a sorting algorithm that uses the divide-and-conquer technique to arrange elements in ascending order.

The algorithm first checks whether the list contains one or no elements. If so, it returns the list because it is already sorted.

If the list has more than one element, it divides it into two smaller halves. It then uses recursion to continue dividing each half until the smaller lists contain only one element.

Finally, it uses the `merge()` function to combine the smaller lists in the correct order until one fully sorted list is produced.

The `assert` statement checks whether the function produces the expected sorted list.

## Section 5 · Quick Sort
Pick a **pivot**. Put everything smaller on its left and everything else on its right. Then quick-sort each side.
This short version builds new lists, so it is easy to read (the in-place version saves memory).

```python
def quick_sort(items):
    if len(items) <= 1:
        return items
    pivot = items[-1]
    rest = items[:-1]
    smaller = [x for x in rest if x < pivot]
    larger = [x for x in rest if x >= pivot]
    return quick_sort(smaller) + [pivot] + quick_sort(larger)

print(quick_sort([7, 2, 1, 6, 8, 5, 3, 4]))     # predict first!
```



Quick Sort is a sorting algorithm that uses the divide-and-conquer technique to arrange elements in ascending order.

In this implementation, the last element is selected as the pivot. The remaining elements are divided into two groups: elements smaller than the pivot and elements greater than or equal to the pivot.

The algorithm then recursively sorts the two groups. Finally, it combines the sorted smaller elements, the pivot, and the sorted larger elements to produce one sorted list.

The base case checks whether the list has one or no elements. If so, it returns the list because it is already sorted.

**Expected output:**
`[1, 2, 3, 4, 5, 6, 7, 8]`

## Section 7 · Sorting Real Records
In real code you rarely write a sort: you call `sorted()` with a **key**. Python's sort is **stable**: records with equal keys keep their original order.

```python
students = [("Brenda", 72), ("Akello", 85), ("Collins", 72), ("Doreen", 91)]

by_mark = sorted(students, key=lambda s: s[1])
print(by_mark)            # who comes first, Brenda or Collins? why?

top_first = sorted(students, key=lambda s: s[1], reverse=True)
print(top_first[0])       # the top student
```

 Sorting Students by Marks

This program sorts students according to their marks using Python's `sorted()` function.

The `key=lambda s: s[1]` tells Python to use the second element in each tuple, which represents the student's marks, when sorting.

The first sorting operation arranges students in ascending order, from the lowest marks to the highest marks.

Brenda comes before Collins because both scored 72 marks, and Python's sorting method is stable. This means students with equal marks maintain their original order.

The second sorting operation uses `reverse=True` to arrange students in descending order, from the highest marks to the lowest marks.

Finally, `top_first[0]` displays the student with the highest marks.
